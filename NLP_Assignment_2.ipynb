```python
#Question 1: Compare and contrast NLTK and spaCy in terms of features, ease of use, and performance.
```

```python
#Answer:

#NLTK (Natural Language Toolkit) and spaCy are popular NLP libraries in Python, but they serve slightly different purposes.
#Features:
#NLTK provides a wide range of NLP tools such as tokenization, stemming, lemmatization, POS tagging, parsing, and access to many linguistic corpora. It is highly flexible and suitable for research and teaching.
#spaCy focuses on industrial-strength NLP with built-in pipelines for tokenization, POS tagging, dependency parsing, named entity recognition (NER), and word vectors.
#Ease of Use:
#NLTK requires more manual setup and multiple steps for tasks, making it better for learning concepts rather than fast deployment.
#spaCy has a simple and clean API with pre-trained models, making it easier and faster to use in real-world applications.
#Performance:
#NLTK is slower and not optimized for large-scale production systems.
#spaCy is highly optimized, written partly in Cython, and performs significantly faster, making it suitable for production environments.
#Conclusion:
#NLTK is ideal for education and experimentation, while spaCy is preferred for high-performance, production-level NLP systems.
```

```python
#Question 4: Describe the architecture and functioning of a Recurrent Natural Network (RNN).
```

```python
#Answer

#A Recurrent Neural Network (RNN) is a type of neural network designed to process sequential data such as text or time-series.
#Architecture:
#Input layer receives sequence data
#Hidden layer maintains a memory (hidden state) of previous inputs
#Output layer generates predictions at each time step
#Functioning:
#At each time step, the hidden state is updated using:
#Current input
#Previous hidden state
#This allows RNNs to capture contextual dependencies in sequences.
#Limitation:
#RNNs suffer from vanishing and exploding gradient problems, making it difficult to learn long-term dependencies.
```

```python
#Question 5: What is the key difference between LSTM and GRU networks in NLP applications?
```

```python
#Answer

#The key difference lies in architecture complexity and gating mechanisms.
#LSTM (Long Short-Term Memory):
#Uses three gates: input, forget, and output
#Has a separate memory cell
#Better at capturing long-term dependencies
#More computationally expensive
#GRU (Gated Recurrent Unit):
#Uses two gates: reset and update
#No separate memory cell
#Faster and simpler than LSTM
#Performs similarly in many NLP tasks
#Conclusion:
#GRUs are efficient and simpler, while LSTMs offer more control over memory handling.
```

```python
##Question 6: Write a Python program using TextBlob to perform sentiment analysis on
#the following paragraph of text:
#“I had a great experience using the new mobile banking app. The interface is intuitive,
#and customer support was quick to resolve my issue. However, the app did crash once
#during a transaction, which was frustrating"
#Your program should print out the polarity and subjectivity scores.
```

```python
#Answer

from textblob import TextBlob

text = """I had a great experience using the new mobile banking app.
The interface is intuitive, and customer support was quick to resolve my issue.
However, the app did crash once during a transaction, which was frustrating."""

blob = TextBlob(text)

print("Polarity:", blob.sentiment.polarity)
print("Subjectivity:", blob.sentiment.subjectivity)
```

```python
#Question 7: Given the sample paragraph below, perform string tokenization and
#frequency distribution using Python and NLTK:
#“Natural Language Processing (NLP) is a fascinating field that combines linguistics,
#computer science, and artificial intelligence. It enables machines to understand,
#interpret, and generate human language. Applications of NLP include chatbots,
#sentiment analysis, and machine translation. As technology advances, the role of NLP
#in modern solutions is becoming increasingly critical.”
```

```python
#Answer

import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

nltk.download('punkt')

text = """Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence."""

tokens = word_tokenize(text)
freq_dist = FreqDist(tokens)

print("Word Frequency Distribution:")
for word, freq in freq_dist.items():
    print(word, ":", freq)

```

```python
#Question 8: Implement a basic LSTM model in Keras for a text classification task using
#the following dummy dataset. Your model should classify sentences as either positive
#(1) or negative (0).
## Dataset
#texts = [
#“I love this project”, #Positive
#“This is an amazing experience”, #Positive
#“I hate waiting in line”, #Negative
#“This is the worst service”, #Negative
#“Absolutely fantastic!” #Positive
#]
#labels = [1, 1, 0, 0, 1]
#Preprocess the text, tokenize it, pad sequences, and build an LSTM model to train on
#this data. You may use Keras with TensorFlow backend.
```

```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = ""
```

```python
#Answer

import os
os.environ["CUDA_VISIBLE_DEVICES"] = ""

import tensorflow as tf


from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

texts = [
    "I love this project",
    "This is an amazing experience",
    "I hate waiting in line",
    "This is the worst service",
    "Absolutely fantastic"
]

labels = np.array([1, 1, 0, 0, 1])

tokenizer = Tokenizer(num_words=100)
tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)
padded = pad_sequences(sequences, maxlen=5)

model = Sequential()
model.add(Embedding(100, 16, input_length=5))
model.add(LSTM(32))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(padded, labels, epochs=10)
```

```python
#Question 9: Using spaCy, build a simple NLP pipeline that includes tokenization,
#lemmatization, and entity recognition. Use the following paragraph as your dataset:
#“Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role in the
#development of India’s atomic energy program. He was the founding director of the Tata
#Institute of Fundamental Research (TIFR) and was instrumental in establishing the
#Atomic Energy Commission of India.”
#Write a Python program that processes this text using spaCy, then prints tokens, their
#lemmas, and any named entities found.
```

```python
#Answer

import torch
import spacy

nlp = spacy.load("en_core_web_sm")

text = """Homi Jehangir Bhaba was an Indian nuclear physicist..."""

doc = nlp(text)

for token in doc:
    print(token.text, token.lemma_)

print("\nNamed Entities:")
for ent in doc.ents:
    print(ent.text, ent.label_)
```
