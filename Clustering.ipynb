# Theory Questions

```python
#1 What is unsupervised learning in the context of machine learning.
```

```python
#Ans: Unsupervised learning finds hidden patterns or structures in unlabeled data (no predefined outputs).
```

```python
#2 How does K-Means work?
```

```python
#Ans: 
#Choose k clusters.
#Randomly initialize centroids.
#Assign points to nearest centroid.
#Update centroids as mean of assigned points.
#Repeat until convergence.
```

```python
#3 Explain the concept of a dendrogram in hierarchical clustering
```

```python
#Ans A tree-like diagram showing how data points merge or split at various distances in hierarchical clustering.
```

```python
#4 Difference between K-Means & Hierarchical Clustering
```

```python
#Ans 
# K-Means: flat, needs k predefined, faster.
# Hierarchical: builds hierarchy, no k needed initially, slower.
```

```python
#5 What are the advantages of DBSCAN over K-Means
```

```python
#Ans 
#No need for k.
#Detects arbitrary-shaped clusters.
#Handles noise/outliers well.
```

```python
#6 When would you use Silhouette Score in clustering
```

```python
#Ans
#To measure clustering quality — use it to evaluate how well samples are assigned to their clusters.
```

```python
#7 What are the limitations of Hierarchical Clustering
```

```python
#Ans
#Computationally expensive (O(n²)).
#Sensitive to noise and outliers.
#Difficult to correct wrong merges/splits.
```

```python
#8  Why is feature scaling important in clustering algorithms like K-Means
```

```python
#Ans 
#Because it uses distance (Euclidean), unscaled features distort cluster formation.
```

```python
#9 How does DBSCAN identify noise points
```

```python
#Ans 
#Points not belonging to any dense region (less than minPts within ε radius) are marked as noise.
```

```python
#10 Define inertia in the context of K-Means
```

```python
#Ans
#Sum of squared distances between each point and its cluster centroid — measures compactness.
```

```python
#11  What is the elbow method in K-Means clustering
```

```python
#Ans
#Plot inertia vs. k; the “elbow” point indicates optimal number of clusters (diminishing improvement after that).
```

```python
#12 Describe the concept of "density" in DBSCAN
```

```python
#Ans Density = number of points within a specified radius (ε). High-density regions form clusters.
```

```python
#13 Can hierarchical clustering be used on categorical data
```

```python
#Ans Yes, but with proper distance measures (e.g., Gower distance, Hamming). Not suitable with Euclidean distance.
```

```python
#14 What does a negative Silhouette Score indicate
```

```python
#Ans A point is likely misclassified — it lies closer to another cluster than its own
```

```python
#15 Explain the term "linkage criteria" in hierarchical clustering
```

```python
#Ans Defines how distances between clusters are computed:
#Single, Complete, Average, Ward linkage.
```

```python
#16Why might K-Means clustering perform poorly on data with varying cluster sizes or densities
```

```python
#Ans Because it assumes equal-sized spherical clusters and equal density — fails on irregular shapes
```

```python
#17 What are the core parameters in DBSCAN, and how do they influence clustering
```

```python
#Ans 
#ε (Eps): neighborhood radius.
#minPts: minimum points for dense region.
#They determine cluster density and noise tolerance.
```

```python
#18 How does K-Means++ improve upon standard K-Means initialization
```

```python
#Ans Chooses initial centroids smartly — spreads them apart to reduce poor clustering and speed up convergence
```

```python
#19 What is agglomerative clustering
```

```python
#Ans A bottom-up hierarchical method — starts with each point as a cluster and merges them iteratively
```

```python
#20 What makes Silhouette Score a better metric than just inertia for model evaluation
```

```python
#Ans Silhouette considers both cohesion (within-cluster) and separation (between-cluster), giving better overall cluster quality
```

```python
#21 Generate synthetic data with 4 centers using make_blobs and apply K-Means clustering. Visualize using a
#scatter plot
```

```python
#Ans 
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=300, centers=4, random_state=42)
kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            s=200, c='red', marker='X')
plt.title("K-Means Clustering (4 Centers)")
plt.show()
```

```python
#29  Create synthetic data using make_blobs and evaluate silhouette scores for k = 2 to 5. Display as a bar chart
```

```python
#Ans
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

X, _ = make_blobs(n_samples=400, centers=4, random_state=42)
scores = []

for k in range(2, 6):
    km = KMeans(n_clusters=k, random_state=42).fit(X)
    scores.append(silhouette_score(X, km.labels_))

plt.bar(range(2, 6), scores, color='skyblue')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Scores for k = 2 to 5")
plt.show()
```

```python
#30  Load the Iris dataset and use hierarchical clustering to group data. Plot a dendrogram with average linkage
```

```python
#Ans 
from sklearn.datasets import load_iris
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

iris = load_iris()
Z = linkage(iris.data, method='average')

plt.figure(figsize=(10, 5))
dendrogram(Z, labels=iris.target)
plt.title("Hierarchical Clustering Dendrogram (Average Linkage)")
plt.xlabel("Sample Index")
plt.ylabel("Distance")
plt.show()
```

```python
#31 Generate synthetic data with overlapping clusters using make_blobs, then apply K-Means and visualize with decision boundaries
```

```python
#Ans 
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import numpy as np

X, _ = make_blobs(n_samples=300, centers=3, cluster_std=2.0, random_state=42)
kmeans = KMeans(n_clusters=3, random_state=42).fit(X)

# Decision boundary
x_min, x_max = X[:, 0].min()-1, X[:, 0].max()+1
y_min, y_max = X[:, 1].min()-1, X[:, 1].max()+1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.05),
                     np.arange(y_min, y_max, 0.05))
Z = kmeans.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.3, cmap='viridis')
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='viridis', s=50)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', marker='X', s=200)
plt.title("K-Means with Decision Boundaries (Overlapping Clusters)")
plt.show()
```

```python
#34   Load the Breast Cancer dataset and compare inertia values for K = 2 to 6 using K-Means. Show results in a line plot
```

```python
#Ans 
from sklearn.datasets import load_breast_cancer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

data = load_breast_cancer().data
inertias = []

for k in range(2, 7):
    kmeans = KMeans(n_clusters=k, random_state=42).fit(data)
    inertias.append(kmeans.inertia_)

plt.plot(range(2, 7), inertias, marker='o')
plt.title("Inertia vs K (Breast Cancer Dataset)")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.show()
```

```python
#46 Load the Wine dataset, apply KMeans, and visualize the cluster assignments in a seaborn pairplot (first 4 features)
```

```python
#Ans 
from sklearn.datasets import load_wine
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

wine = load_wine()
df = pd.DataFrame(wine.data, columns=wine.feature_names)
df_scaled = pd.DataFrame(StandardScaler().fit_transform(df), columns=wine.feature_names)

kmeans = KMeans(n_clusters=3, random_state=42).fit(df_scaled)
df_scaled['Cluster'] = kmeans.labels_

sns.pairplot(df_scaled.iloc[:, :5], hue='Cluster', palette='Set2')
plt.show()
```
